```python
import matplotlib.pyplot as plt

days_of_week = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
temperatures = [20, 22, 23, 21, 24, 26, 25]

plt.figure(figsize=(10, 6))
plt.plot(days_of_week, temperatures, marker='o', linestyle='-', color='skyblue')

plt.xlabel('Day of the Week')
plt.ylabel('Temperature (°C)')
plt.title('Temperature Variation Over a Week')

plt.grid(True, linestyle='--', alpha=0.7)


plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset
# The dataset typically has 'Date' and 'Temp' columns.
# Ensure the 'Date' column is parsed as datetime objects.
# Using the existing file '1_Daily_minimum_temps.csv' found in the environment.
# Explicitly specifying date_format to avoid UserWarning about inferring format.
df = pd.read_csv('1_Daily_minimum_temps.csv', header=0, index_col=0, parse_dates=True, date_format='%Y-%m-%d')

# Select the first 7 days
df_first_7_days = df.head(7)

# Create the line plot
plt.figure(figsize=(10, 6))
plt.plot(df_first_7_days.index, df_first_7_days['Temp'], marker='o', linestyle='-', color='coral')

# Add labels and title
plt.xlabel('Date')
plt.ylabel('Temperature (°C)')
plt.title('Daily Minimum Temperatures for the First 7 Days')

# Add grid for better readability
plt.grid(True, linestyle='--', alpha=0.7)

# Display the plot
plt.show()
```
